### (a)

The ELU activation function with $\alpha = 0.1$ is

$$
\operatorname{ELU}(x)=
\begin{cases}
0.1(e^x-1), & x<0,\\
x, & x\geq 0.
\end{cases}
$$

### Output Range

For $x<0$,

$$
e^x \in (0,1)
$$

so,

$$
e^x-1 \in (-1,0)
$$

and therefore,

$$
0.1(e^x-1)\in(-0.1,0).
$$

For $x\geq 0$,

$$
\operatorname{ELU}(x)=x\in[0,\infty).
$$

Therefore, the output range of ELU is

$$
\boxed{\operatorname{ELU}(x)\in(-0.1,\infty)}.
$$

### Derivative

For $x<0$,

$$
\frac{d}{dx}\left[0.1(e^x-1)\right]
=0.1e^x.
$$

For $x>0$,

$$
\frac{d}{dx}[x]=1.
$$

Hence,

$$
\operatorname{ELU}'(x)=
\begin{cases}
0.1e^x, & x<0,\\
1, & x>0.
\end{cases}
$$

At $x=0$,

$$
\lim_{x\to0^-}\operatorname{ELU}'(x)=0.1
$$

and

$$
\lim_{x\to0^+}\operatorname{ELU}'(x)=1.
$$

Since

$$
0.1\neq1,
$$

the derivative is not defined at $x=0$.

Therefore,

$$
\boxed{
\operatorname{ELU}'(x)=
\begin{cases}
0.1e^x, & x<0,\\
\text{undefined}, & x=0,\\
1, & x>0.
\end{cases}
}
$$

---

### (b)

The GELU activation function is

$$
\operatorname{GELU}(x)=x\Phi(x),
$$

where $\Phi(x)$ is the cumulative distribution function of the standard Gaussian distribution.

Also,

$$
\Phi'(x)=\phi(x),
$$

where

$$
\phi(x)=\frac{1}{\sqrt{2\pi}}e^{-x^2/2}.
$$

### Derivative

Using the product rule,

$$
\frac{d}{dx}[x\Phi(x)]
=
\frac{d}{dx}[x]\Phi(x)
+
x\frac{d}{dx}[\Phi(x)].
$$

Therefore,

$$
\operatorname{GELU}'(x)
=
\Phi(x)+x\phi(x).
$$

Substituting the Gaussian probability density function,

$$
\boxed{
\operatorname{GELU}'(x)
=
\Phi(x)
+
\frac{x}{\sqrt{2\pi}}e^{-x^2/2}
}
$$

### Output Range

As $x\to\infty$,

$$
\Phi(x)\to1.
$$

Therefore,

$$
\operatorname{GELU}(x)=x\Phi(x)\to\infty.
$$

For negative $x$, GELU can produce a small negative value. Its minimum value is approximately

$$
\operatorname{GELU}(x)\approx-0.17.
$$

Therefore, the output range is approximately

$$
\boxed{
\operatorname{GELU}(x)\in[-0.17,\infty)
}.
$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import ndtr

x = np.linspace(-5, 5, 1000)

elu = np.where(x < 0, 0.1 * (np.exp(x) - 1), x)

x_neg = x[x < 0]
x_pos = x[x > 0]

phi = (1 / np.sqrt(2 * np.pi)) * np.exp(-(x ** 2) / 2)
Phi = ndtr(x)

gelu = x * Phi
gelu_derivative = Phi + x * phi

plt.figure(figsize=(8, 5))

plt.plot(x, elu, label='ELU(x)')
plt.plot(x_neg, 0.1 * np.exp(x_neg), label="ELU'(x), x < 0")
plt.plot(x_pos, np.ones_like(x_pos), label="ELU'(x), x > 0")
plt.axhline(0, linewidth=0.5)
plt.axvline(0, linewidth=0.5)
plt.xlabel('x')
plt.ylabel('Value')
plt.title('ELU and its Derivative (a = 0.1)')
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(x, gelu, label='GELU(x)')
plt.plot(x, gelu_derivative, label="GELU'(x)")
plt.axhline(0, linewidth=0.5)
plt.axvline(0, linewidth=0.5)
plt.xlabel('x')
plt.ylabel('Value')
plt.title('GELU Activation Function and Derivative')
plt.legend()
plt.grid()
plt.show()
```

### (g) Cross-Entropy Loss for a General Classification Problem

For a classification problem with $M$ classes, the cross-entropy loss is

$$
CE(y,p(x))
=
-\sum_{i=1}^{M}y_i\log p_i,
$$

where $y$ is the one-hot vector representing the true class.

For example, if there are four classes and the true class is $y=2$, then

$$
y=
\begin{bmatrix}
0 & 1 & 0 & 0
\end{bmatrix}.
$$

Therefore,

$$
CE(y,p(x))
=
-
\left(
0\log p_1
+
1\log p_2
+
0\log p_3
+
0\log p_4
\right).
$$

Hence,

$$
CE(y,p(x))
=
-\log p_2.
$$

In general, cross-entropy only uses the predicted probability of the correct class, so

$$
\boxed{
CE(y,p(x))=-\log p_y
}.
$$

Since probabilities are between $0$ and $1$,

$$
0<p_y\leq1.
$$

Therefore,

$$
\log p_y\leq0.
$$

Hence,

$$
-\log p_y\geq0.
$$

Therefore, the cross-entropy loss is always non-negative:

$$
\boxed{
CE(y,p(x))\geq0
}.
$$

The loss becomes zero when

$$
-\log p_y=0,
$$

which occurs when

$$
p_y=1,
$$

because

$$
-\log(1)=0.
$$

```python
class MyLinear(torch.nn.Module):
  def __init__(self, input_size, output_size):
    """
    input_size: the size of the input
    output_size: the size of the output
    """
    super().__init__()
    #Your code here
    self.W = torch.nn.Parameter(torch.randn(input_size, output_size) * 0.01)
    self.b = torch.nn.Parameter(torch.zeros(output_size))

  #forward propagation
  def forward(self, x): #x is a mini-batch
    #Your code here
    return x @ self.W + self.b
```

```python
class MyFFN(torch.nn.Module):
  def __init__(self, input_size, num_classes, hidden_sizes, act = torch.nn.ReLU()):
    """
    input_size: the size of the input
    num_classes: the number of classes
    act is the activation function
    hidden_sizes is the list of hidden sizes
    for example input_size = 3, hidden_sizes = [5, 7], num_classes = 4, and act = torch.nn.ReLU()
    means that we are building up a FFN with the confirguration
    (3 (Input) -> 5 (ReLU) -> 7 (ReLU) -> 4 (Output))
    """
    super(MyFFN, self).__init__()
    self.input_size = input_size
    self.num_classes = num_classes
    self.act = act
    self.hidden_sizes = hidden_sizes
    self.num_layers = len(hidden_sizes) + 1

  def create_FFN(self):
    """
    This function creates the feed-forward neural network
    We stack many MyLinear layers
    """
    hidden_sizes = [self.input_size] + self.hidden_sizes + [self.num_classes]
    self.layers = []
    #Your code here
    for i in range(self.num_layers):
      self.layers.append(MyLinear(hidden_sizes[i], hidden_sizes[i+1]))

  def forward(self,x):
    """
    This implements the forward propagation of the batch x
    This needs to return the logits of x
    """
    #Your code here
    activation = self.act() if isinstance(self.act, type) else self.act

    for i in range(self.num_layers):
      x = self.layers[i](x)
      if i < self.num_layers - 1:
        x = activation(x)
    return x

  def compute_loss(self, x, y):
    """
    This function computes the cross-entropy loss
    You can use the built-in CE loss of PyTorch
    """
    #Your code here
    logits = self.forward(x)
    loss = torch.nn.functional.cross_entropy(logits, y)
    return loss

  def update_SGD(self, x, y, learning_rate = 0.01):
    """
    This function updates the model parameters using SGD using the batch (x,y)
    You need to implement the update rule manually and cannot rely on the built-in optimizer
    """
    #Your code here
    for layer in self.layers:
      if layer.W.grad is not None:
        layer.W.grad.zero_()
      if layer.b.grad is not None:
        layer.b.grad.zero_()

    loss = self.compute_loss(x, y)
    loss.backward()

    with torch.no_grad():
      for layer in self.layers:
        layer.W -= learning_rate * layer.W.grad
        layer.b -= learning_rate * layer.b.grad

  def update_SGDwithMomentum(self, x, y, learning_rate = 0.01, momentum = 0.9):
    """
    This function updates the model parameters using SGD with momentum using the batch (x,y)
    You need to implement the update rule manually and cannot rely on the built-in optimizer
    """
    #Your code here
    for layer in self.layers:
      if layer.W.grad is not None:
        layer.W.grad.zero_()
      if layer.b.grad is not None:
        layer.b.grad.zero_()

    loss = self.compute_loss(x, y)
    loss.backward()

    if not hasattr(self,'velocity_W'):
      self.velocity_W = [
          torch.zeros_like(layer.W)
          for layer in self.layers
      ]
      self.velocity_b = [
          torch.zeros_like(layer.b)
          for layer in self.layers
      ]
    with torch.no_grad():
      for i, layer in enumerate(self.layers):
        self.velocity_W[i] = momentum * layer.W.grad + (1 - momentum) * self.velocity_W[i]
        self.velocity_b[i] = momentum * layer.b.grad + (1 - momentum) * self.velocity_b[i]
        layer.W -= learning_rate * self.velocity_W[i]
        layer.b -= learning_rate * self.velocity_b[i]

  def update_AdaGrad(self, x, y, learning_rate = 0.01):
    """
    This function updates the model parameters using AdaGrad using the batch (x,y)
    You need to implement the update rule manually and cannot rely on the built-in optimizer
    """
    #Your code here
    for layer in self.layers:
      if layer.W.grad is not None:
        layer.W.grad.zero_()
      if layer.b.grad is not None:
        layer.b.grad.zero_()

    loss = self.compute_loss(x, y)
    loss.backward()

    if not hasattr(self, 'gamma_W'):
      self.gamma_W = [
          torch.zeros_like(layer.W)
          for layer in self.layers
      ]
      self.gamma_b = [
          torch.zeros_like(layer.b)
          for layer in self.layers
      ]

    epsilon = 1e-6

    with torch.no_grad():
      for i, layer in enumerate(self.layers):
        self.gamma_W[i] += layer.W.grad ** 2
        self.gamma_b[i] += layer.b.grad ** 2

        layer.W -= (
            learning_rate * layer.W.grad
            / (torch.sqrt(self.gamma_W[i]) + epsilon)
        )

        layer.b -= (
            learning_rate * layer.b.grad
            / (torch.sqrt(self.gamma_b[i]) + epsilon)
        )
```

**(c)** Write the code to evaluate the accuracy of the current `myFFN` model on a data loader (e.g., train_loader or test_loader).

<div style="text-align: right"><font color="red">[2.5 points]</font></div>

```python
def compute_acc(model, data_loader):
  """
  This function computes the accuracy of the model on a data loader
  """
  #Your code here
  correct = 0
  total = 0

  with torch.no_grad():
    for x, y in data_loader:
      logits = model(x)
      predictions = torch.argmax(logits, dim=1)
      correct += torch.sum(predictions == y).item()
      total += y.size(0)

  accuracy = correct / total
  return accuracy
```

```python
def compute_loss(model, data_loader):
  """
  This function computes the loss of the model on a data loader
  """
  #Your code here
  total_loss = 0
  total_samples = 0

  with torch.no_grad():
    for x, y in data_loader:
      loss = model.compute_loss(x, y)
      total_loss += loss.item() * y.size(0)
      total_samples += y.size(0)

  average_loss = total_loss / total_samples
  return average_loss
```
